**Import All librariesq Required**


```python
#suppress warnings
def warn(*args, **kwargs):
    pass
import warnings
warnings.warn = warn
warnings.filterwarnings('ignore')

#Findspark simplifieds the pocess of using Apache Spark with python
import findspark
findspark.init()

#Importq SparkSession
from pyspark.sql import SparkSession

```

Task-1 Create Spark Session

```python
#Create SparkSession
#Here 'Getting Started with Spark' is the application name
#Ignore any warnings by SparkSession command

spark = SparkSession.builder.appName("Getting Sarted with Spark").getOrCreate()
```

**Tas-2 Download Dta File**

**Task 3 - Load the data in a csv file into a dataframe**

```python
# using the spark.read.csv function we load the data into a dataframe.
# the header = True mentions that there is a header row in out csv file
# the inferSchema = True, tells spark to automatically find out the data types of the columns.

#load mpg dataset
mpg_data = spark.read.csv("mpg.csv", header=True, inferSchema=True)
```

**Task 4 - Explore the data set**

Let's print the schema of the dataset

```python
mpg_data.printSchema()
```

```python
# show top 5 rows from the dataset
mpg_data.head(10)
```

```python
spark.stop()
```
